# Results viewer

Builds `outputs/results_viewer.html`, one page for every run in `outputs/`. Open it in a browser (it reads the images from the run folders, so keep it in `outputs/`).

- **Plots**: pick a result from the Mechanical Solution tree (deformation, core/hardpoint shear, von Mises, contact status/pressure, interaction indices) and see it for every run that passes the filters. Group the plots by an input (e.g. core thickness), pick a `results.csv` value to print under each plot, and sort by it. Click a plot to enlarge it; the arrow keys step through the runs.
- **Table**: every column of `results.csv` (inputs, pass/fail, margins, stresses and allowables, deformation, contact pressure, bolt and joint forces, weights, run info), filtered the same way. Toggle column groups, click a header to sort, click a row to see its plot.
- **Filters** (left): tick or untick the values of each input that varies between runs. Inputs that are the same in every run are listed below them.

Plots are plan views (front edge up) at the end of load step 2, in the units of `result_units`; layer results are at the middle of the core/hardpoint layer. **Each plot has its own colour scale**: compare the legends, or the values under the plots, not the colours. Contact pressure legends are capped at `contact_plot_max`.

New runs export their plots when they solve, and the sweep rewrites the page after every run. The first cell is only needed for runs solved before the plots were exported automatically.

```python
%load_ext autoreload
%autoreload 2

import logging
import webbrowser

from pedal_floor import viewer

logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s", force=True)
logging.getLogger("ansys").setLevel(logging.WARNING)
```

Export the plots of saved runs that have none yet: one Mechanical session opens each `model.mechdb` in turn (read only) and takes about 2 s per run. Runs open in another Mechanical window are skipped; close it and run the cell again. `overwrite=True` redoes every run.

```python
outcome = viewer.export_images(overwrite=False)
for run, result in outcome.items():
    if not result.endswith("images"):
        print(run, result)
print(f"{len(outcome)} runs processed")
```

```python
page = viewer.build()
print(page)
webbrowser.open(page.as_uri())
```
